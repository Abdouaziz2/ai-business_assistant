# Partie 1 — Anatomie d’un prompt

## 1. Problème

> Je souhaite analyser les retours de clients d'une entreprise.

L'objectif est de construire un prompt permettant à une IA d'analyser plusieurs commentaires clients de manière structurée.

---

## 2. Prompt utilisé

Tu es un analyste spécialisé dans l'analyse des retours clients.

Analyse les commentaires clients que je vais te fournir.

Pour chaque commentaire, identifie :

* le sentiment exprimé : positif, négatif ou neutre ;
* les problèmes rencontrés par le client ;
* les points positifs éventuels ;
* les recommandations que l'entreprise pourrait prendre en compte.

Présente les résultats sous forme de tableau avec les colonnes suivantes :

| Commentaire | Sentiment | Problèmes | Points positifs | Recommandations |

Contraintes :

* base-toi uniquement sur les informations présentes dans les commentaires ;
* n'invente aucune information ;
* si une information n'est pas présente, indique « Non précisé » ;
* utilise un langage clair et professionnel.

Commentaires à analyser :

« Le service est rapide et le personnel est très sympathique. »

« J'ai attendu trois jours avant de recevoir ma commande. »

« L'application est intéressante mais elle plante régulièrement. »

---

## 3. Résultat obtenu

L'IA a identifié les éléments suivants :

| Commentaire                                                    | Sentiment | Problèmes                          | Points positifs                       | Recommandations                         |
| -------------------------------------------------------------- | --------- | ---------------------------------- | ------------------------------------- | --------------------------------------- |
| Le service est rapide et le personnel est très sympathique.    | Positif   | Non précisé                        | Service rapide, personnel sympathique | Non précisé                             |
| J'ai attendu trois jours avant de recevoir ma commande.        | Négatif   | Délai de livraison de trois jours  | Non précisé                           | Améliorer le délai de livraison         |
| L'application est intéressante mais elle plante régulièrement. | Négatif   | L'application plante régulièrement | Application intéressante              | Améliorer la stabilité de l'application |

---

## 4. Anatomie du prompt

| Composante   | Élément utilisé                                                     | Objectif                            |
| ------------ | ------------------------------------------------------------------- | ----------------------------------- |
| Rôle         | Analyste spécialisé dans les retours clients                        | Donner un contexte précis à l'IA    |
| Tâche        | Analyser les commentaires clients                                   | Définir l'action à réaliser         |
| Instructions | Identifier sentiment, problèmes, points positifs et recommandations | Décomposer la tâche                 |
| Format       | Tableau avec cinq colonnes                                          | Obtenir une réponse structurée      |
| Contraintes  | Ne pas inventer et utiliser « Non précisé »                         | Limiter les réponses incorrectes    |
| Données      | Trois commentaires clients                                          | Fournir les informations à analyser |
---
## 5. Analyse

Le prompt produit une réponse structurée et directement exploitable.
La présence d'un rôle permet de préciser le contexte de la tâche. Les instructions détaillent les informations recherchées. Le tableau impose une structure commune aux différents commentaires.

Les contraintes permettent également de limiter l'invention d'informations et de gérer les informations absentes.
---

## 6. Ce que j'ai appris

Un prompt peut être construit à partir de plusieurs composantes :

**Rôle + Tâche + Instructions + Format + Contraintes + Données**

Plus les instructions sont précises, plus il est possible de contrôler la réponse obtenue.

Cette première partie montre donc qu'un prompt ne consiste pas simplement à poser une question à une IA : il consiste à lui donner des instructions suffisamment précises pour obtenir le résultat attendu.
