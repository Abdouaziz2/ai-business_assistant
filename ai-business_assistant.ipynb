# Partie 1 — Anatomie d’un prompt

## 1. Problème

> Je souhaite analyser les retours de clients d'une entreprise.

L'objectif est de construire un prompt permettant à une IA d'analyser plusieurs commentaires clients de manière structurée.

---

## 2. Prompt utilisé

Tu es un analyste spécialisé dans l'analyse des retours clients.

Analyse les commentaires clients que je vais te fournir.

Pour chaque commentaire, identifie :

* le sentiment exprimé : positif, négatif ou neutre ;
* les problèmes rencontrés par le client ;
* les points positifs éventuels ;
* les recommandations que l'entreprise pourrait prendre en compte.

Présente les résultats sous forme de tableau avec les colonnes suivantes :

| Commentaire | Sentiment | Problèmes | Points positifs | Recommandations |

Contraintes :

* base-toi uniquement sur les informations présentes dans les commentaires ;
* n'invente aucune information ;
* si une information n'est pas présente, indique « Non précisé » ;
* utilise un langage clair et professionnel.

Commentaires à analyser :

« Le service est rapide et le personnel est très sympathique. »

« J'ai attendu trois jours avant de recevoir ma commande. »

« L'application est intéressante mais elle plante régulièrement. »

---

## 3. Résultat obtenu

L'IA a identifié les éléments suivants :

| Commentaire                                                    | Sentiment | Problèmes                          | Points positifs                       | Recommandations                         |
| -------------------------------------------------------------- | --------- | ---------------------------------- | ------------------------------------- | --------------------------------------- |
| Le service est rapide et le personnel est très sympathique.    | Positif   | Non précisé                        | Service rapide, personnel sympathique | Non précisé                             |
| J'ai attendu trois jours avant de recevoir ma commande.        | Négatif   | Délai de livraison de trois jours  | Non précisé                           | Améliorer le délai de livraison         |
| L'application est intéressante mais elle plante régulièrement. | Négatif   | L'application plante régulièrement | Application intéressante              | Améliorer la stabilité de l'application |

---

## 4. Anatomie du prompt

| Composante   | Élément utilisé                                                     | Objectif                            |
| ------------ | ------------------------------------------------------------------- | ----------------------------------- |
| Rôle         | Analyste spécialisé dans les retours clients                        | Donner un contexte précis à l'IA    |
| Tâche        | Analyser les commentaires clients                                   | Définir l'action à réaliser         |
| Instructions | Identifier sentiment, problèmes, points positifs et recommandations | Décomposer la tâche                 |
| Format       | Tableau avec cinq colonnes                                          | Obtenir une réponse structurée      |
| Contraintes  | Ne pas inventer et utiliser « Non précisé »                         | Limiter les réponses incorrectes    |
| Données      | Trois commentaires clients                                          | Fournir les informations à analyser |
---
## 5. Analyse

Le prompt produit une réponse structurée et directement exploitable.
La présence d'un rôle permet de préciser le contexte de la tâche. Les instructions détaillent les informations recherchées. Le tableau impose une structure commune aux différents commentaires.

Les contraintes permettent également de limiter l'invention d'informations et de gérer les informations absentes.
---

## 6. Ce que j'ai appris

Un prompt peut être construit à partir de plusieurs composantes :

**Rôle + Tâche + Instructions + Format + Contraintes + Données**

Plus les instructions sont précises, plus il est possible de contrôler la réponse obtenue.

Cette première partie montre donc qu'un prompt ne consiste pas simplement à poser une question à une IA : il consiste à lui donner des instructions suffisamment précises pour obtenir le résultat attendu.


# Partie 2 — Zero-shot, One-shot, Few-shot et prompt structuré

## 1. Objectif

L'objectif est de comparer différentes techniques de prompting sur une même tâche de classification.

### Commentaire à analyser

> « Le service est rapide mais l'application plante régulièrement. »

### Catégories possibles

* positif
* négatif
* neutre

---

## 2. Zero-shot

### Principe

Le Zero-shot consiste à demander au modèle d'effectuer une tâche sans lui fournir d'exemple.

### Prompt

> Classe le commentaire suivant dans une seule des catégories suivantes : positif, négatif ou neutre.
>
> Commentaire :
> « Le service est rapide mais l'application plante régulièrement. »
>
> Retourne uniquement la catégorie.

### Résultat

```text
négatif
```

### Observation

Le modèle a correctement choisi la catégorie négative sans avoir reçu d'exemple.

---

## 3. One-shot

### Principe

Le One-shot consiste à fournir un seul exemple avant de demander au modèle de réaliser la tâche.

### Exemple fourni

> « Le produit est excellent et le service est très satisfaisant. » → positif

### Résultat

```text
négatif
```

### Observation

Le résultat est identique à celui obtenu avec le Zero-shot.

---

## 4. Few-shot

### Principe

Le Few-shot consiste à fournir plusieurs exemples afin de montrer au modèle comment réaliser la tâche.

### Exemples fournis

> « Le produit est excellent et le service est très satisfaisant. » → positif
>
> « Le produit est inutilisable et le service client ne répond jamais. » → négatif
>
> « Le magasin est ouvert aujourd'hui et propose plusieurs produits. » → neutre

### Résultat

```text
négatif
```

### Observation

Le modèle donne encore la même classification.

---

## 5. Prompt structuré

### Principe

Le prompt structuré organise les instructions en différentes parties afin de rendre la demande plus précise.

Dans notre test, le prompt contient :

* un contexte ;
* une tâche ;
* les catégories ;
* des règles de classification ;
* le commentaire ;
* un format de sortie précis.

### Résultat

```text
négatif
```

### Observation

Le résultat reste identique, mais les instructions sont beaucoup plus explicites et permettent notamment de préciser comment traiter un commentaire contenant à la fois des éléments positifs et négatifs.

---

## 6. Comparaison

| Technique        | Nombre d'exemples | Résultat |
| ---------------- | ----------------: | -------- |
| Zero-shot        |                 0 | négatif  |
| One-shot         |                 1 | négatif  |
| Few-shot         |                 3 | négatif  |
| Prompt structuré |                 0 | négatif  |

---
Dans cette expérience, les quatre techniques ont produit la même classification : **négatif**.

L'expérience montre que :

* le Zero-shot permet de réaliser une tâche sans exemple ;
* le One-shot montre au modèle un exemple ;
* le Few-shot montre plusieurs exemples ;
* le prompt structuré permet d'organiser précisément les instructions et les règles.

Dans notre cas particulier, l'ajout d'exemples n'a pas changé le résultat final. Le commentaire contient un élément positif concernant la rapidité du service, mais le problème récurrent de l'application est considéré comme dominant, ce qui conduit à la catégorie négative.



# Partie 3 — Prompt Engineering et raisonnement

## 3.1 Décomposer un prompt

### Prompt de départ

> « Analyse ces avis clients et donne-moi les problèmes les plus importants ainsi que les recommandations. »

Ce prompt contient plusieurs éléments implicites :

* **Tâche :** analyser des avis clients.
* **Informations à analyser :** les avis clients fournis.
* **Objectif :** identifier les problèmes importants.
* **Sortie attendue :** donner également des recommandations.
* **Critère d’importance :** non précisé.
* **Format de réponse :** non précisé.
* **Contraintes :** non précisées.

### Limites du prompt

Le prompt est compréhensible, mais il manque plusieurs informations pour obtenir une réponse plus précise et reproductible.

Par exemple, il ne précise pas :

* comment déterminer l’importance d’un problème ;
* comment regrouper les problèmes similaires ;
* comment traiter les informations absentes ;
* sous quelle forme présenter les résultats ;
* s’il faut éviter les interprétations et les informations non présentes dans les commentaires.

L’objectif du Prompt Engineering est donc de transformer une demande générale en consignes plus précises.

---

## 3.2 Analyse avec contraintes

### Prompt utilisé

```text
Analyse les commentaires clients suivants.

Identifie :

1. les principaux problèmes rencontrés ;
2. les points positifs ;
3. les recommandations possibles pour l'entreprise.

Regroupe les problèmes similaires et classe-les par importance.

Commentaires :

- « La livraison est arrivée avec deux jours de retard. »
- « Le produit fonctionne parfaitement. »
- « Le service client ne répond pas à mes messages. »
- « La livraison était rapide mais l'emballage était abîmé. »
```

### Résultat obtenu

```text
### 1. Principaux problèmes rencontrés

| Importance | Problème | Commentaires concernés | Justification |
|------------|----------|------------------------|---------------|
| 1 — Élevée | Retards de livraison | « La livraison est arrivée avec deux jours de retard. » | Le client signale explicitement un retard de deux jours. |
| 2 — Élevée | Absence de réponse du service client | « Le service client ne répond pas à mes messages. » | Le client indique clairement qu'il n'obtient pas de réponse. |
| 3 — Moyenne | Emballage abîmé | « La livraison était rapide mais l'emballage était abîmé. » | Le problème concerne l'état de l'emballage à la réception. |

### Problèmes similaires regroupés :

- Livraison : retard de deux jours + problème d'emballage.
- Service client : absence de réponse aux messages.

### 2. Points positifs

- Le produit fonctionne parfaitement.
- Une livraison est décrite comme rapide.

### 3. Recommandations possibles

1. Améliorer la ponctualité des livraisons afin de réduire les retards.
2. Améliorer la réactivité du service client et assurer un suivi des messages.
3. Renforcer la qualité de l'emballage pour éviter qu'il soit abîmé pendant la livraison.
```

---

## 3.3 Vérification du raisonnement

### Pourquoi vérifier une réponse ?

Une réponse peut être **cohérente et bien présentée** tout en contenant des éléments qui ne sont pas directement justifiés par les données.

Il faut donc vérifier la réponse produite par le modèle.

Le but n'est pas simplement de demander :

> « Est-ce que ta réponse est correcte ? »

Il faut donner au modèle des critères précis de vérification.

### Prompt de vérification utilisé

```text
Vérifie ton analyse précédente à partir des quatre commentaires clients fournis.

Pour chaque élément de ton analyse, vérifie s'il est directement justifié par les commentaires.

Identifie :

1. les affirmations qui sont correctement justifiées ;
2. les affirmations qui sont des généralisations ou des interprétations ;
3. les informations qui ne sont pas présentes dans les commentaires ;
4. les éventuelles contradictions ;
5. les recommandations qui pourraient être mieux justifiées.

Ne modifie pas les commentaires d'origine et n'invente aucune information.

Termine par une conclusion indiquant si l'analyse initiale est fiable ou si elle doit être corrigée.

Commentaires d'origine :

- « La livraison est arrivée avec deux jours de retard. »
- « Le produit fonctionne parfaitement. »
- « Le service client ne répond pas à mes messages. »
- « La livraison était rapide mais l'emballage était abîmé. »

Analyse précédente :

[COLLE ICI LA RÉPONSE OBTENUE À L'ÉTAPE PRÉCÉDENTE]
```

### Résultat obtenu

#### 1. Affirmations correctement justifiées

**Retards de livraison — correctement justifié.**

Le commentaire « La livraison est arrivée avec deux jours de retard » justifie directement l'identification d'un **retard de livraison**.

**Absence de réponse du service client — correctement justifié.**

« Le service client ne répond pas à mes messages » justifie directement le problème d'**absence de réponse du service client**.

**Emballage abîmé — correctement justifié.**

« La livraison était rapide mais l'emballage était abîmé » justifie directement l'identification d'un **emballage abîmé**.

**Points positifs — correctement justifiés.**

* « Le produit fonctionne parfaitement » justifie que le produit fonctionne bien.
* « La livraison était rapide » justifie que, dans ce commentaire précis, la livraison a été rapide.

#### 2. Généralisations ou interprétations

La principale interprétation problématique concerne le **classement par importance** :

* « 1 — Élevée : Retards de livraison »
* « 2 — Élevée : Absence de réponse du service client »
* « 3 — Moyenne : Emballage abîmé »

Les commentaires ne fournissent **aucun critère permettant de mesurer l'importance** de ces problèmes.

On peut les identifier, mais on ne peut pas affirmer objectivement que le retard est plus important que l'absence de réponse du service client.

Autre interprétation :

> « Problèmes similaires regroupés : Livraison : retard de deux jours + problème d'emballage. »

Ces deux problèmes concernent effectivement la **livraison**, mais ils ne sont pas exactement le même problème.

Il serait plus précis de les regrouper sous une catégorie générale comme :

**« Problèmes liés à la livraison »**

tout en conservant deux problèmes distincts.

Enfin :

> « les incidents signalés par certains clients »

Cette formulation peut être remplacée par :

> « les commentaires fournis signalent ces problèmes »

afin de rester strictement basé sur les données disponibles.

#### 3. Informations absentes des commentaires

Certaines formulations des recommandations ajoutent des éléments qui ne sont pas explicitement présents :

* « assurer un suivi des messages » : le commentaire indique seulement que le service client ne répond pas ;
* « renforcer la qualité de l'emballage » : l'emballage est signalé comme abîmé, mais la cause de cette détérioration n'est pas connue ;
* « améliorer la ponctualité des livraisons » : c'est une recommandation logique à partir du retard, mais ce n'est pas une information présente dans le commentaire.

Ces éléments sont donc des **recommandations déduites**, et non des faits observés.

#### 4. Contradictions éventuelles

Il n'y a **pas de contradiction directe** dans l'analyse.

Cependant, les deux commentaires concernant la livraison montrent des expériences différentes :

* une livraison est arrivée avec **deux jours de retard** ;
* une autre livraison était **rapide**, mais l'emballage était **abîmé**.

Cela ne constitue pas une contradiction, car il s'agit de commentaires différents.

On ne peut simplement pas conclure que les livraisons sont toujours lentes ou toujours rapides.

#### 5. Recommandations à mieux justifier

Les recommandations sont globalement pertinentes, mais elles devraient être formulées de manière plus directement liée aux faits :

1. **Retard :** surveiller et réduire les délais de livraison, puisque deux jours de retard sont explicitement signalés.
2. **Service client :** améliorer la réactivité aux messages, puisque l'absence de réponse est explicitement signalée.
3. **Emballage :** vérifier et améliorer la protection des colis, puisque l'emballage est arrivé abîmé.

Il faut éviter d'ajouter des causes non démontrées, par exemple affirmer que **le transport** est responsable de l'emballage abîmé.

### Conclusion de la vérification

**L'analyse initiale est globalement fiable, mais elle doit être légèrement corrigée.**

Les **problèmes et points positifs sont correctement identifiés**.

En revanche, le **classement par importance n'est pas directement justifié par les commentaires**, et certaines recommandations vont légèrement au-delà des informations disponibles.

La version la plus rigoureuse consiste donc à distinguer clairement :

> **faits observés → interprétations éventuelles → recommandations déduites**

# 3.4 Améliorer le prompt

La vérification précédente permet d'identifier une faiblesse importante :

> Le prompt demande de classer les problèmes « par importance », mais ne définit pas ce que signifie « importance ».

Pour rendre le résultat plus reproductible, il faut donc préciser le critère.

### Prompt amélioré

```text
# Contexte

Nous souhaitons analyser des retours clients afin d'identifier les problèmes rencontrés par les clients et les actions possibles pour l'entreprise.

# Tâche

Analyse les commentaires clients fournis.

Identifie :

1. les principaux problèmes rencontrés ;
2. les points positifs ;
3. les recommandations possibles pour l'entreprise.

# Règles

- Base-toi uniquement sur les informations présentes dans les commentaires.
- N'invente aucune information.
- Ne suppose pas la cause d'un problème si elle n'est pas indiquée.
- Regroupe les problèmes qui appartiennent à une même catégorie générale, mais conserve les problèmes distincts séparément.
- Pour chaque problème, indique les commentaires qui le justifient.
- Si une information n'est pas présente, indique « Non précisé ».
- Les recommandations doivent être directement liées aux problèmes observés.
- Ne transforme pas une recommandation en fait.

# Importance

Classe les problèmes selon les critères suivants :

- Élevée : problème ayant un impact directement important sur l'expérience du client ou empêchant l'utilisation normale du service ou du produit.
- Moyenne : problème gênant l'expérience du client mais ne bloquant pas complètement le service.
- Faible : problème mineur ou ayant un impact limité.

Si les commentaires ne permettent pas de déterminer objectivement l'importance d'un problème, indique « Importance non déterminable » au lieu d'inventer un classement.

# Format

Présente les résultats sous forme de trois sections :

1. Problèmes rencontrés
2. Points positifs
3. Recommandations

Pour les problèmes, utilise le tableau :

| Importance | Catégorie | Problème | Commentaire concerné | Justification |

Commentaires :

- « La livraison est arrivée avec deux jours de retard. »
- « Le produit fonctionne parfaitement. »
- « Le service client ne répond pas à mes messages. »
- « La livraison était rapide mais l'emballage était abîmé. »
```

### Pourquoi ce prompt est meilleur ?

Le prompt initial demandait simplement :

> « classe-les par importance »

Le prompt amélioré explique **comment déterminer l'importance**.

Il ajoute également plusieurs garde-fous :

* source des informations ;
* absence d'invention ;
* distinction entre faits et recommandations ;
* regroupement contrôlé ;
* justification de chaque problème ;
* gestion des informations absentes ;
* critère explicite d'importance.

On obtient donc un prompt **plus précis, plus contrôlable et plus reproductible**.

# 3.5 Comparaison : prompt initial vs prompt amélioré

| Élément               | Prompt initial            | Prompt amélioré              |
| --------------------- | ------------------------- | ---------------------------- |
| Tâche                 | Analyser les commentaires | Analyser les commentaires    |
| Problèmes             | Demandés                  | Demandés + justifiés         |
| Points positifs       | Demandés                  | Demandés                     |
| Recommandations       | Demandées                 | Demandées et liées aux faits |
| Regroupement          | Demandé                   | Défini précisément           |
| Importance            | Non définie               | Critères définis             |
| Informations absentes | Non précisé               | « Non précisé »              |
| Invention             | Non précisé               | Interdite explicitement      |
| Causes supposées      | Non précisé               | Interdites                   |
| Format                | Peu détaillé              | Tableau précis               |
| Vérification          | Absente                   | Possible                     |
| Reproductibilité      | Moyenne                   | Meilleure                    |

# 3.6 Faits, interprétations et recommandations

Un point essentiel du Prompt Engineering est de ne pas mélanger différentes catégories d'informations.

## 1. Fait observé

C'est une information directement présente dans les données.

Exemple :

> « Le service client ne répond pas à mes messages. »

Fait observé :

> Le client indique que le service client ne répond pas à ses messages.

## 2. Interprétation

C'est une conclusion déduite à partir d'un fait.

Exemple :

> Le service client est probablement mal organisé.

Cette information **n'est pas présente** dans le commentaire.

Il faut donc éviter de la présenter comme un fait.

## 3. Recommandation

C'est une action proposée à partir des problèmes identifiés.

Exemple :

> Améliorer la réactivité du service client.

Cette information n'est pas un fait présent dans le commentaire. C'est une **proposition**.

### Schéma à retenir

```text
DONNÉES
   ↓
FAITS OBSERVÉS
   ↓
INTERPRÉTATIONS
   ↓
RECOMMANDATIONS
```

Plus on descend dans cette chaîne, plus le risque d'ajouter une interprétation augmente.

Il faut donc toujours pouvoir revenir aux données originales.

---

# 3.7 Vérification finale d'une réponse

Pour contrôler une réponse générée par une IA, on peut utiliser cette checklist :

### Vérification des faits

* [ ] Chaque affirmation est-elle présente dans les données ?
* [ ] Une information a-t-elle été inventée ?
* [ ] Une cause a-t-elle été supposée sans preuve ?
* [ ] Les commentaires originaux ont-ils été respectés ?

### Vérification des interprétations

* [ ] Certaines conclusions vont-elles au-delà des données ?
* [ ] Y a-t-il des généralisations ?
* [ ] Le modèle présente-t-il une interprétation comme un fait ?
* [ ] Les contradictions apparentes ont-elles été vérifiées ?

### Vérification des recommandations

* [ ] Chaque recommandation correspond-elle à un problème identifié ?
* [ ] La recommandation ajoute-t-elle une information non démontrée ?
* [ ] Peut-on expliquer pourquoi cette recommandation a été proposée ?

### Vérification du format

* [ ] Le format demandé a-t-il été respecté ?
* [ ] Toutes les informations demandées sont-elles présentes ?
* [ ] Les informations absentes sont-elles correctement indiquées ?

# 3.8 Exercice pratique

## Objectif

Construire un prompt permettant d'analyser automatiquement les avis suivants.

### Données

```text
- « Le personnel est très accueillant. »
- « J'ai reçu ma commande avec trois jours de retard. »
- « Le produit est de bonne qualité mais l'application est difficile à utiliser. »
- « Le service client a répondu rapidement à ma demande. »
- « L'emballage était endommagé à la réception. »
```

### Consigne

Construis un prompt qui demande à l'IA de :

1. identifier les problèmes ;
2. identifier les points positifs ;
3. regrouper les problèmes similaires ;
4. indiquer leur importance ;
5. proposer des recommandations ;
6. justifier chaque élément avec le commentaire concerné ;
7. ne rien inventer.

### Contrainte supplémentaire

Le prompt doit préciser ce que signifie :

* **importance élevée** ;
* **importance moyenne** ;
* **importance faible**.

---

# 3.9 Exercice de vérification

Après avoir obtenu la réponse de l'IA, utilise le prompt suivant :

```text
Vérifie l'analyse précédente.

Pour chaque élément, indique :

1. ce qui est directement présent dans les commentaires ;
2. ce qui est une interprétation ;
3. ce qui est une recommandation ;
4. ce qui n'est pas justifié par les données ;
5. les éventuelles généralisations ;
6. les éventuelles contradictions.

Ne modifie pas les commentaires d'origine.

Pour chaque élément non justifié, explique pourquoi.

Termine par :

- « Analyse fiable »
ou
- « Analyse à corriger »

Justifie ta conclusion.
```

---

# 3.10 Ce que démontre l'exercice

Cet exercice montre une idée importante du Prompt Engineering :

> **Un modèle peut produire une réponse cohérente sans que chaque élément de cette réponse soit directement justifié par les données fournies.**

La deuxième étape permet donc de **vérifier la première réponse**.

Le raisonnement consiste à séparer :

### 1. Faits observés

→ Ce qui est explicitement présent dans les commentaires.

### 2. Interprétations

→ Ce que le modèle déduit à partir des commentaires.

### 3. Recommandations

→ Les actions proposées à partir des problèmes identifiés.

### 4. Informations non justifiées

→ Les éléments ajoutés sans preuve dans les données.

Cette vérification permet de détecter notamment :

* les généralisations ;
* les interprétations excessives ;
* les contradictions ;
* les informations inventées ou non présentes ;
* les recommandations insuffisamment justifiées.

---

# 3.11 Principe général à retenir

Le Prompt Engineering ne consiste donc pas seulement à **obtenir une réponse**.

Il consiste également à **contrôler la qualité de cette réponse**.

Un bon processus peut être résumé ainsi :

```text
1. FORMULER
      ↓
2. PRODUIRE
      ↓
3. VÉRIFIER
      ↓
4. CORRIGER
      ↓
5. VALIDER
```

### Principe à retenir

> **Produire → vérifier → corriger → valider**

# Partie 4 — Sorties structurées

## 4.1 — Transformer une réponse textuelle en JSON

### Contexte

Le LLM retourne initialement une réponse textuelle :

> « Le commentaire semble plutôt négatif. Le client est mécontent du délai de livraison... »

Cette réponse est compréhensible par un humain, mais une application aura plus de facilité à exploiter une structure JSON.

### Prompt utilisé

```text
Tu es un système d'analyse automatique de commentaires clients.

Analyse le commentaire suivant et retourne uniquement un objet JSON valide.

Le JSON doit contenir exactement les champs suivants :

- "sentiment" : chaîne de caractères. Valeurs autorisées : "positif", "negatif", "neutre".
- "categorie" : chaîne de caractères. Valeurs autorisées : "livraison", "produit", "service_client", "autre".
- "urgence" : chaîne de caractères. Valeurs autorisées : "faible", "moyenne", "élevée".
- "probleme" : chaîne de caractères décrivant brièvement le problème identifié.
- "confiance" : nombre décimal compris entre 0 et 1.

Ne retourne aucun texte avant ou après le JSON.
N'ajoute aucune propriété supplémentaire.

Commentaire :
« Le client est mécontent car sa commande a été livrée avec deux jours de retard. »
```

### Résultat obtenu

```json
{
  "sentiment": "negatif",
  "categorie": "livraison",
  "urgence": "moyenne",
  "probleme": "Commande livrée avec deux jours de retard.",
  "confiance": 0.99
}
```

### Analyse du résultat

| Champ       | Type attendu        | Valeur obtenue                                 | Conforme |
| ----------- | ------------------- | ---------------------------------------------- | -------- |
| `sentiment` | string              | `"negatif"`                                    | ✅        |
| `categorie` | string              | `"livraison"`                                  | ✅        |
| `urgence`   | string              | `"moyenne"`                                    | ✅        |
| `probleme`  | string              | `"Commande livrée avec deux jours de retard."` | ✅        |
| `confiance` | number entre 0 et 1 | `0.99`                                         | ✅        |

### Conclusion

Le prompt permet de transformer une réponse textuelle en une **sortie structurée**, composée de champs clairement définis.

### Prompt utilisé

```text
Tu es un système d'analyse automatique de commentaires clients.

Analyse le commentaire fourni et retourne uniquement un objet JSON valide.

Le JSON doit respecter strictement les règles suivantes :

1. Le format doit être un JSON valide.

2. Le JSON doit contenir exactement les 5 propriétés suivantes :
   - "sentiment"
   - "categorie"
   - "urgence"
   - "probleme"
   - "confiance"

3. Aucune propriété supplémentaire n'est autorisée.

4. "sentiment" doit être une chaîne de caractères et sa valeur doit être exactement l'une des suivantes :
   - "positif"
   - "negatif"
   - "neutre"

5. "categorie" doit être une chaîne de caractères et sa valeur doit être exactement l'une des suivantes :
   - "livraison"
   - "produit"
   - "service_client"
   - "autre"

6. "urgence" doit être une chaîne de caractères et sa valeur doit être exactement l'une des suivantes :
   - "faible"
   - "moyenne"
   - "élevée"

7. "probleme" doit être une chaîne de caractères.

8. "confiance" doit être un nombre compris entre 0 et 1 inclus.

9. Ne retourne aucun commentaire, aucune explication et aucun texte en dehors du JSON.

10. Respecte exactement les noms des propriétés indiqués.

Commentaire :
« Le client est mécontent car sa commande a été livrée avec deux jours de retard. »
```

## 4.3 — Résumé de la Partie 4

```text
Prompt
   ↓
LLM
   ↓
JSON structuré
   ↓
Validation des règles
   ↓
Sortie exploitable par l'application
```

### Ce qu'on a appris

Un prompt structuré permet de contrôler non seulement **le contenu de la réponse**, mais également **sa forme**.

Il faut définir :

* les propriétés attendues ;
* le type de chaque propriété ;
* les valeurs autorisées ;
* les contraintes numériques ;
* l'interdiction de propriétés supplémentaires ;
* le format final attendu.

### Principe à retenir

> **Une sortie structurée est un contrat entre le LLM et l'application.**

Le LLM produit les données, tandis que les règles de validation permettent de vérifier que ces données respectent le format attendu.
